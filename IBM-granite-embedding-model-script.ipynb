```python
from huggingface_hub import login, HfApi

# Replace with your actual Write token
login(token="YOUR_HUGGING_FACE_WRITE_TOKEN")  # Replace with your actual token
```

```python
api = HfApi()

# Define your repo name (username / desired-repo-name)
repo_id = "arshvir/granite-embedding-311m-multilingual-r2-GGUF"

# Create the repository (it will be public by default)
api.create_repo(repo_id=repo_id, exist_ok=True, repo_type="model")

# Upload the GGUF file from the /content/ folder
print("Uploading model to Hugging Face...")
api.upload_file(
    path_or_fileobj="/content/granite-311m-multilingual-r2-f16.gguf",
    path_in_repo="granite-311m-multilingual-r2-f16.gguf",
    repo_id=repo_id,
    repo_type="model"
)
print(f"Upload complete! View your model at: https://huggingface.co/{repo_id}")
```
